```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder, OneHotEncoder
import warnings
warnings.filterwarnings('ignore')

# Baca dataset
df = pd.read_csv('/content/drive/MyDrive/Semester 5/Praktikum Penambangan Data/Pertemuan 2/Bengaluru_House_Data.csv')
df
```

```python
# 2. Membuat Histogram dan Penjelasan
# Function helper untuk convert total_sqft
def convert_sqft_to_num(x):
    if pd.isna(x):
        return np.nan
    if '-' in str(x):
        try:
            values = str(x).split('-')
            return (float(values[0]) + float(values[1])) / 2
        except:
            return np.nan
    try:
        return float(x)
    except:
        return np.nan

# Membuat histogram
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Histogram Analysis - Bengaluru House Price Data', fontsize=16, fontweight='bold')

# Histogram Price
if 'price' in df.columns:
    axes[0,0].hist(df['price'].dropna(), bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    axes[0,0].set_title('Distribution of House Prices')
    axes[0,0].set_xlabel('Price (in lakhs)')
    axes[0,0].set_ylabel('Frequency')

# Histogram Size (BHK)
if 'size' in df.columns:
    size_numeric = pd.to_numeric(df['size'].str.extract('(\d+)')[0], errors='coerce')
    axes[0,1].hist(size_numeric.dropna(), bins=15, alpha=0.7, color='lightgreen', edgecolor='black')
    axes[0,1].set_title('Distribution of House Size (BHK)')
    axes[0,1].set_xlabel('Number of Bedrooms')
    axes[0,1].set_ylabel('Frequency')

# Histogram Total sqft
if 'total_sqft' in df.columns:
    sqft_numeric = df['total_sqft'].apply(convert_sqft_to_num)
    axes[1,0].hist(sqft_numeric.dropna(), bins=30, alpha=0.7, color='salmon', edgecolor='black')
    axes[1,0].set_title('Distribution of Total Square Feet')
    axes[1,0].set_xlabel('Total Square Feet')
    axes[1,0].set_ylabel('Frequency')

# Histogram Bath
if 'bath' in df.columns:
    axes[1,1].hist(df['bath'].dropna(), bins=15, alpha=0.7, color='gold', edgecolor='black')
    axes[1,1].set_title('Distribution of Bathrooms')
    axes[1,1].set_xlabel('Number of Bathrooms')
    axes[1,1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
## 3. Korelasi Antar Variabel dan Penjelasan
# Siapkan data untuk korelasi
df_corr = df.copy()

# Convert categorical ke numeric
if 'size' in df_corr.columns:
    df_corr['size_numeric'] = pd.to_numeric(df_corr['size'].str.extract('(\d+)')[0], errors='coerce')

if 'total_sqft' in df_corr.columns:
    df_corr['total_sqft_numeric'] = df_corr['total_sqft'].apply(convert_sqft_to_num)

# Hitung korelasi hanya untuk kolom numerik
numeric_cols = df_corr.select_dtypes(include=[np.number]).columns
correlation_matrix = df_corr[numeric_cols].corr()

# Tampilkan correlation matrix
correlation_matrix

# Visualisasi heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
            square=True, linewidths=0.5, fmt='.2f')
plt.title('Correlation Matrix - Bengaluru House Price Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
## 4. Membuat Descriptive Analysis dan Penjelasan

# Descriptive statistics untuk numerik
df.describe()

# Analisis kategorikal
categorical_cols = df.select_dtypes(include=['object']).columns
for col in categorical_cols:
    print(f"\n{col} - Top 10 categories:")
    print(df[col].value_counts().head(10))
    print(f"Total unique values: {df[col].nunique()}")
```

```python
## 5. Cek Missing Value dan Penanganan

# Check missing values
missing_data = df.isnull().sum()
missing_percent = (missing_data / len(df)) * 100
missing_df = pd.DataFrame({
    'Column': missing_data.index,
    'Missing_Count': missing_data.values,
    'Missing_Percentage': missing_percent.values
})
missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Percentage', ascending=False)
missing_df

# Visualisasi missing values
if len(missing_df) > 0:
    plt.figure(figsize=(12, 6))
    plt.bar(missing_df['Column'], missing_df['Missing_Percentage'], color='coral', alpha=0.7)
    plt.title('Missing Values Percentage by Column')
    plt.xlabel('Columns')
    plt.ylabel('Missing Percentage (%)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

# Imputation
df_cleaned = df.copy()
for col in df_cleaned.columns:
    if df_cleaned[col].isnull().any():
        if df_cleaned[col].dtype == 'object':
            # Categorical: isi dengan mode
            mode_val = df_cleaned[col].value_counts()
            if len(mode_val) > 0:
                most_frequent = mode_val.index[0]
                df_cleaned[col].fillna(most_frequent, inplace=True)
            else:
                df_cleaned[col].fillna('Unknown', inplace=True)
        else:
            # Numerical: isi dengan median
            median_val = df_cleaned[col].median()
            df_cleaned[col].fillna(median_val, inplace=True)

# Check hasil imputation
df_cleaned.isnull().sum()
```

```python
## 6. Cek Duplikasi dan Penanganan

# Check duplicates
duplicate_count = df_cleaned.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")

if duplicate_count > 0:
    print(f"Percentage of duplicates: {(duplicate_count/len(df_cleaned))*100:.2f}%")

    # Tampilkan sample duplicates
    duplicates = df_cleaned[df_cleaned.duplicated(keep=False)].sort_values(df_cleaned.columns.tolist())
    duplicates.head(10)

    # Remove duplicates
    df_cleaned.drop_duplicates(inplace=True)
    print(f"After removing duplicates: {len(df_cleaned)} rows")
else:
    print("No duplicate rows found!")

df_cleaned.shape
```

```python
## 7. Encoding untuk Categorical Variabel

# Siapkan data
df_X = df_cleaned.drop(['price'], axis=1, errors='ignore')
df_y = df_cleaned['price'] if 'price' in df_cleaned.columns else None

# Identifikasi categorical columns
cats = df_X.select_dtypes(include=['object', 'bool']).columns
print("Categorical columns:", list(cats))

# 7a. ONE HOT ENCODING - Style 1
df_onehot = pd.get_dummies(df_X, columns=cats, drop_first=True)
df_onehot.head()

# 7b. ONE HOT ENCODING - Style 2
ohe = OneHotEncoder(sparse_output=False, drop='first')
ohe.fit(df_X[cats])
df_ohe_array = ohe.transform(df_X[cats])
print("OneHotEncoder shape:", df_ohe_array.shape)
df_ohe_array[:5]

# 7c. LABEL ENCODING
df_X_label = df_X.copy()
label_encoders = {}

for col in cats:
    le = LabelEncoder()
    df_X_label[col] = le.fit_transform(df_X_label[col].astype(str))
    label_encoders[col] = le

df_X_label.head()
```

```python
## 8. Pisah Dataset menjadi Train Test (80:20)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    df_X_label, df_y,
    test_size=0.2,    # 80:20 split
    random_state=42   # for reproducibility
)

print(f"Training set shape: {X_train.shape}")
print(f"Test set shape: {X_test.shape}")
print(f"Training target shape: {y_train.shape}")
print(f"Test target shape: {y_test.shape}")

# Tampilkan sample
X_train.head()
```

```python
## 9. Standardization dan Normalization

# 9a. STANDARDIZATION
scaler_std = StandardScaler()
scaler_std.fit(X_train)  # Fit hanya pada training data
X_train_std = scaler_std.transform(X_train)
X_test_std = scaler_std.transform(X_test)

print("After Standardization:")
print("Training set shape:", X_train_std.shape)
print("Sample standardized data:")
print(X_train_std[:5])

# Verificasi standardization
print(f"Mean: {np.mean(X_train_std, axis=0)[:5]}")
print(f"Std: {np.std(X_train_std, axis=0)[:5]}")

# 9b. NORMALIZATION
scaler_norm = MinMaxScaler()
scaler_norm.fit(X_train)  # Fit hanya pada training data
X_train_norm = scaler_norm.transform(X_train)
X_test_norm = scaler_norm.transform(X_test)

print("\nAfter Normalization:")
print("Training set shape:", X_train_norm.shape)
print("Sample normalized data:")
print(X_train_norm[:5])

# Verificasi normalization
print(f"Min: {np.min(X_train_norm, axis=0)[:5]}")
print(f"Max: {np.max(X_train_norm, axis=0)[:5]}")
```

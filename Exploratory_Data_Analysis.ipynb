# EDA (Exploratory Data Analysis)

Exploratory Data Analysis (EDA) merupakan proses analisis awal yang dilakukan untuk memahami dataset yang dimiliki sebelum menjalankan model statistik atau melakukan pengambilan keputusan.

**Deskripsi Data:** EDA dimulai dengan meringkas karakteristik utama dari dataset yang termasuk statistik deskriptif seperti mean, median, modus, kuartil, dan sebagainya. Ini membantu dalam pemahaman awal tentang distribusi dan struktur data.

**Visualisasi Data:** Penggunaan grafik dan plot adalah bagian integral dari EDA. Visualisasi memungkinkan untuk melihat pola, tren, dan anomali dalam data dengan cara yang lebih intuitif.

**Identifikasi Anomali:** EDA membantu dalam mendeteksi nilai-nilai anomali dalam dataset.

**Pemrosesan Data:** Dalam proses EDA, seringkali diperlukan untuk membersihkan dan mempersiapkan data sebelum analisis lebih lanjut.

**Pemeriksaan Korelasi:** EDA membantu dalam memahami hubungan antara variabel dalam dataset.

**Pemilihan Fitur:** Melalui EDA, dapat dipilih fitur-fitur yang paling relevan atau memiliki pengaruh signifikan terhadap variabel target.

**Eksplorasi Lebih Lanjut:** EDA dapat menjadi titik awal untuk pertanyaan lebih lanjut dan penelitian lanjutan. Analisis awal dapat mengungkap pola atau tren yang menarik, yang kemudian dapat diinvestigasi lebih lanjut untuk mendapatkan pemahaman yang lebih dalam tentang data.

**Interpretasi Hasil:** Selain dari mempersiapkan data untuk analisis, EDA juga membantu dalam menafsirkan hasil dari analisis yang dilakukan. Ini membantu pengambilan keputusan yang lebih baik berdasarkan pemahaman yang mendalam tentang data yang ada.

### bar plot
Bar plot adalah jenis grafik yang menampilkan nilai-nilai kategori dalam satu variabel dengan menggunakan batang vertikal atau horizontal untuk mewakili nilai atau frekuensi setiap kategori, memudahkan perbandingan antara kategori-kategori tersebut.

### stacked bar plot
Stacked bar plot adalah jenis grafik yang menunjukkan komposisi variabel dalam satu kategori dengan menggunakan batang vertikal atau horizontal yang ditumpuk satu sama lain untuk mewakili kontribusi relatif dari setiap variabel pada kategori tersebut, sehingga memungkinkan untuk melihat hubungan antara variabel-variabel tersebut dalam satu kategori.

### Sub plot

Subplot adalah fitur dalam visualisasi data yang memungkinkan pengguna untuk membuat beberapa grafik atau plot secara bersamaan dalam satu gambar atau "figure", memudahkan perbandingan atau representasi visual dari beberapa dataset atau hasil analisis secara paralel.







```python
#versi 1
import matplotlib.pyplot as plt
import numpy as np

#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1) #1 baris, 2 kolom, dan plot ini adalah plot pertama.
plt.plot(x,y)
plt.title('title 1')

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 40, 20])

plt.subplot(1, 2, 2) #1 baris, 2 kolom, dan plot ini adalah plot kedua.
plt.plot(x,y)
plt.title('title 2')

plt.show()
```

## Summary

Penjelasan penggunaan grafik dalam visualisasi data sebagai berikut:

**Distribusi (Distribution):**

Histogram: Digunakan untuk menampilkan distribusi frekuensi atau probabilitas dari data numerik. Setiap batang dalam histogram mewakili interval data, dan tinggi batang menunjukkan frekuensi atau probabilitas kejadian di interval tersebut.

Scatter plot: Digunakan untuk menampilkan distribusi data titik dan menampilkan pola atau hubungan antara dua variabel numerik. Setiap titik dalam scatter plot mewakili satu pengamatan dan posisinya ditentukan oleh nilai dari dua variabel yang berkaitan.

**Hubungan (Relationship):**

Scatter plot: Digunakan untuk mengeksplorasi hubungan antara dua variabel numerik. Dengan menggambar titik-titik pada bidang kartesian, scatter plot membantu dalam menemukan pola atau korelasi antara variabel-variabel tersebut.

**Perbandingan (Comparison):**

Bar plot: Digunakan untuk membandingkan nilai-nilai kategori dalam satu variabel. Setiap batang dalam bar plot mewakili satu kategori dan tingginya menunjukkan nilai atau frekuensi kategori tersebut.

Line plot: Digunakan untuk membandingkan tren atau perubahan nilai variabel numerik dari waktu ke waktu. Garis yang menghubungkan titik-titik data membantu dalam menyoroti pola atau tren dari data tersebut.

**Komposisi (Composition)**

Pie chart: Digunakan untuk menunjukkan komposisi relatif dari bagian-bagian. Setiap "slice" dari pie chart mewakili proporsi atau persentase dari keseluruhan.

Stacked bar chart: Digunakan untuk menunjukkan komposisi variabel dalam satu kategori. Setiap batang dalam stacked bar chart dibagi menjadi segmen-segmen yang mewakili kontribusi dari variabel-variabel tersebut pada kategori tersebut.

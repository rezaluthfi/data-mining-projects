```python
df_new = df.drop('CUST_ID', axis = 1)
df_new.head()
```

```python
df_new.isnull().sum()
```

```python
df_new['MINIMUM_PAYMENTS'].fillna(df_new['MINIMUM_PAYMENTS'].median(), inplace=True)
df_new['CREDIT_LIMIT'].fillna(df_new['CREDIT_LIMIT'].median(), inplace=True)
```

```python
df_new.isnull().sum()
```

```python
from sklearn.preprocessing import StandardScaler
#from sklearn.preprocessing import MinMaxScaler
X = df_new.astype(float).values
scaler = StandardScaler().fit(X)
X_new= scaler.transform(X)
X_new
```

```python
from sklearn.cluster import KMeans
```

```python
inertia_list = []
for num_clusters in range(1, 11):
 kmeans_model = KMeans(n_clusters=num_clusters)
 kmeans_model.fit(X_new)
 inertia_list.append(kmeans_model.inertia_)
 print("For n_clusters = {}, inertia value is {})".format(num_clusters, kmeans_model.inertia_))
```

```python
plt.plot(range(1,11),inertia_list, marker='o', linewidth=2, markersize=8)
plt.xlabel("Number of Clusters", size=13)
plt.ylabel("Inertia Value", size=13)
plt.title("Inertia values vary depending on the number of clusters utilized.")
plt.show()
```

```python
 from sklearn.metrics import silhouette_samples, silhouette_score

 sh_list = []
 for num_clusters in range(2, 11):
  kmeans = KMeans(n_clusters=num_clusters)
  cluster_labels = kmeans.fit_predict(X_new)
  score = silhouette_score(X_new, cluster_labels)
  sh_list.append(score)

  print("For n_clusters = {}, silhouette score is {})".format(num_clusters, score))
```

```python
plt.plot(range(2,11),sh_list, marker='o', linewidth=2, markersize=8)
plt.xlabel("Number of Clusters", size=13)
plt.ylabel("silhouette score", size=13)
plt.title("Silhoutte score values vary depending on the number of clusters utilized.")
plt.show()
# di stage ini kita menemukan k=3 dengan score tertinggi
```

```python
k_means = KMeans(n_clusters = 3, random_state = 42)
k_means.fit(X_new)
labels = k_means.labels_
df_new['cluster_labels'] = labels
df_new.head()
```

```python
# check centroids-nya
centroids = k_means.cluster_centers_
centroids
```

```python
x1 = df_new['PURCHASES']
x2 = df_new['PAYMENTS']

plt.figure(figsize=(8,6))
u_labels = np.unique(labels)

for i in u_labels:
 plt.scatter(x1[df_new['cluster_labels'] == i] , x2[df_new['cluster_labels'] == i] , label = i)

plt.scatter(x1,x2, c=k_means.labels_, cmap='rainbow')
plt.xlabel(x1.name,  fontsize=20)
plt.ylabel(x2.name,  fontsize=20)
plt.title('K-means clustering',fontsize=20)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)

plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn_extra.cluster import KMedoids
from sklearn.metrics import silhouette_score

from google.colab import drive
drive.mount('/content/drive')

file_path = '/content/drive/My Drive/Semester 5/Praktikum Penambangan Data/Pertemuan 8/Air_Traffic_Passenger_Statistics_20251113.csv'

try:
    df = pd.read_csv(file_path)
    print("Dataset berhasil dimuat.")
except FileNotFoundError:
    print(f"Error: File tidak ditemukan di path '{file_path}'. Pastikan path sudah benar.")

df.head()
```

```python
# Menampilkan informasi dasar dataset
print("Informasi Dataset:")
df.info()

print("\nStatistik Deskriptif untuk Kolom Numerik:")
print(df.describe())

# Menghitung jumlah nilai unik di kolom-kolom penting
print("\nJumlah Nilai Unik:")
print(f"Jumlah Maskapai Operasi Unik: {df['Operating Airline'].nunique()}")
print(f"Jumlah Rangkuman GEO Unik: {df['GEO Summary'].nunique()}")
print(f"Jumlah Wilayah GEO Unik: {df['GEO Region'].nunique()}")
```

```python
# Pembersihan Data dan Konversi Tipe Data
# Kolom 'Passenger Count' saat ini bertipe 'object' karena ada koma.

print(f"Tipe data 'Passenger Count' sebelum dibersihkan: {df['Passenger Count'].dtype}")

# Hapus karakter koma dari kolom 'Passenger Count'
# .str.replace() hanya bisa digunakan pada kolom string (object)
df['Passenger Count'] = df['Passenger Count'].str.replace(',', '', regex=False)

# Ubah tipe data kolom menjadi numerik (integer)
# pd.to_numeric akan mengonversi string angka menjadi angka
df['Passenger Count'] = pd.to_numeric(df['Passenger Count'])

print(f"Tipe data 'Passenger Count' setelah dibersihkan: {df['Passenger Count'].dtype}")
print("\nPembersihan data selesai. Tipe data sudah benar.")

# Verifikasi dengan melihat info
df.info()
```

```python
# Membuat profil maskapai berdasarkan total dan rata-rata penumpang
airline_profile = df.groupby('Operating Airline')['Passenger Count'].agg(['sum', 'mean', 'count']).reset_index()
airline_profile.columns = ['Operating Airline', 'Total Passengers', 'Avg Passengers per Entry', 'Activity Record Count']

# Membuat pivot table untuk memisahkan jumlah penumpang domestik dan internasional
passenger_geo = pd.pivot_table(
    df,
    values='Passenger Count',
    index='Operating Airline',
    columns='GEO Summary',
    aggfunc='sum',
    fill_value=0
)

# Menggabungkan semua fitur menjadi satu DataFrame profil maskapai
df_final = pd.merge(airline_profile, passenger_geo, on='Operating Airline', how='left')

# Mengisi nilai NaN jika ada (seharusnya tidak ada karena fill_value=0)
df_final = df_final.fillna(0)

print("\nProfil Maskapai yang Telah Dibuat:")
df_final.head()

# Memisahkan fitur numerik untuk penskalaan
features = df_final.drop('Operating Airline', axis=1)

# Penskalaan Fitur
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

print("\nDimensi data setelah penskalaan:", features_scaled.shape)
```

```python
# Menentukan jumlah cluster optimal menggunakan K-Means
inertia_list = []
silhouette_list = []
k_range = range(2, 15) # Mencoba k dari 2 hingga 14

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(features_scaled)
    inertia_list.append(kmeans.inertia_)
    silhouette_list.append(silhouette_score(features_scaled, kmeans.labels_))

# Plotting Elbow Method
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(k_range, inertia_list, marker='o', linestyle='--')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method untuk K-Means')
plt.grid(True)

# Plotting Silhouette Score
plt.subplot(1, 2, 2)
plt.plot(k_range, silhouette_list, marker='o', linestyle='--')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score untuk K-Means')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
k_optimal = 2

# Menerapkan K-Means dengan k optimal
kmeans = KMeans(n_clusters=k_optimal, random_state=42, n_init=10)
df_final['Cluster_KMeans'] = kmeans.fit_predict(features_scaled)

# Menganalisis hasil cluster
kmeans_summary = df_final.groupby('Cluster_KMeans').mean(numeric_only=True)
print("Ringkasan Rata-Rata per Cluster (K-Means):")
print(kmeans_summary)

# Visualisasi Hasil K-Means
plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=df_final,
    x='Total Passengers',
    y='International', # Menggunakan jumlah penumpang internasional sebagai sumbu y
    hue='Cluster_KMeans',
    palette='viridis',
    s=100, # Ukuran titik
    alpha=0.8
)
plt.title('Hasil Clustering K-Means Maskapai Penerbangan (k=2)')
plt.xlabel('Total Penumpang (Skala Log)')
plt.ylabel('Total Penumpang Internasional')
plt.xscale('log') # Menggunakan skala log untuk visualisasi yang lebih baik
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```

```python
# Menerapkan K-Medoids dengan k optimal yang sama
kmedoids = KMedoids(n_clusters=k_optimal, random_state=42)
df_final['Cluster_KMedoids'] = kmedoids.fit_predict(features_scaled)

# Menganalisis hasil cluster K-Medoids
kmedoids_summary = df_final.groupby('Cluster_KMedoids').mean(numeric_only=True)
print("\nRingkasan Rata-Rata per Cluster (K-Medoids):")
print(kmedoids_summary)

# Visualisasi Hasil K-Medoids
plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=df_final,
    x='Total Passengers',
    y='International',
    hue='Cluster_KMedoids',
    palette='plasma',
    s=100,
    alpha=0.8
)
plt.title('Hasil Clustering K-Medoids Maskapai Penerbangan')
plt.xlabel('Total Penumpang (Skala Log)')
plt.ylabel('Total Penumpang Internasional')
plt.xscale('log')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```
